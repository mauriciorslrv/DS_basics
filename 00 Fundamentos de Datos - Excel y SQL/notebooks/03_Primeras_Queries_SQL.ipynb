**Antes de comenzar:** una query es una pregunta que escribimos para consultar una tabla. Esta notebook funciona sin instalar PostgreSQL porque usa SQLite, incorporado en Python. Más adelante repetirás la lógica en PostgreSQL.

**Objetivo:** leer siete ventas ficticias, hacer consultas pequeñas y entender qué hace cada palabra clave.

```python
import sqlite3
import pandas as pd
conexion = sqlite3.connect(":memory:")
conexion.executescript("""
CREATE TABLE clientes (cliente_id INTEGER PRIMARY KEY,nombre TEXT,pais TEXT);
CREATE TABLE productos (producto_id INTEGER PRIMARY KEY,nombre TEXT,precio REAL);
CREATE TABLE ventas (venta_id INTEGER PRIMARY KEY,cliente_id INTEGER,producto_id INTEGER,cantidad INTEGER,fecha TEXT);
INSERT INTO clientes VALUES (1,'Ana','MX'),(2,'Bruno','US'),(3,'Carla','CA'),(4,'Diego','MX'),(5,'Eva','US');
INSERT INTO productos VALUES (1,'Albahaca',28),(2,'Romero',32),(3,'Menta',26);
INSERT INTO ventas VALUES (1,1,1,10,'2025-01-06'),(2,2,2,5,'2025-01-12'),(3,1,3,8,'2025-02-02'),(4,3,1,7,'2025-02-13'),(5,4,2,6,'2025-03-11'),(6,2,1,12,'2025-03-19'),(7,5,3,3,'2025-03-22');
""")
def consultar(sql):
    return pd.read_sql_query(sql,conexion)
assert len(consultar("SELECT * FROM ventas")) == 7
print("¡Listo! Tenemos 5 clientes, 3 productos y 7 ventas ficticias.")
```

## 1 · SELECT y FROM: ¿qué información quiero ver?

`SELECT` elige las columnas que quieres mostrar y `FROM` indica de qué tabla salen.

Lee la consulta antes de ejecutarla. ¿Cuántas columnas esperas obtener?

```python
consultar("SELECT nombre, pais FROM clientes")
```

**Tu turno:** modifica `nombre, pais` por `nombre`. ¿Qué cambia? Después intenta `SELECT * FROM productos`: el asterisco pide todas las columnas.

## 2 · WHERE: ¿qué filas me interesan?

`WHERE` filtra los registros que cumplen una condición. Las comillas se usan para textos.

```python
consultar("SELECT nombre, pais FROM clientes WHERE pais = 'MX'")
```

**Reto:** cambia `MX` por `CA`, luego prueba `US`. ¿Cuántas personas encuentras de cada país?

## 3 · ORDER BY: ¿en qué orden quiero el resultado?

`ORDER BY` clasifica filas; `DESC` indica orden descendente.

```python
consultar("SELECT nombre, precio FROM productos ORDER BY precio DESC")
```

## 4 · COUNT y SUM: ¿cuánto hay?

`COUNT(*)` cuenta filas. `SUM(cantidad)` suma las unidades registradas; cantidad no es lo mismo que número de operaciones.

```python
consultar("SELECT COUNT(*) AS operaciones, SUM(cantidad) AS unidades FROM ventas")
```

## 5 · GROUP BY: ¿cuánto hay por grupo?

Con `GROUP BY` resumes por categoría. Aquí preguntamos: ¿cuántas compras realizó cada cliente?

```python
consultar("SELECT cliente_id, COUNT(*) AS compras, SUM(cantidad) AS unidades FROM ventas GROUP BY cliente_id ORDER BY compras DESC")
```

## 6 · JOIN: ¿quién compró qué?

Las ventas guardan identificadores. `JOIN` relaciona ventas con clientes y productos, para convertirlos en nombres legibles.

```python
consultar("""
SELECT c.nombre AS cliente, p.nombre AS producto, v.cantidad,
       ROUND(v.cantidad * p.precio, 2) AS ingreso
FROM ventas v
JOIN clientes c ON v.cliente_id = c.cliente_id
JOIN productos p ON v.producto_id = p.producto_id
ORDER BY ingreso DESC
""")
```
