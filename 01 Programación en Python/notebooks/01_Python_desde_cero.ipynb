## 1. Programar es dividir un problema

Una computadora no adivina lo que queremos. Necesita instrucciones claras y en un orden específico.

Imagina una compra: cada producto cuesta 120 pesos, compras 3 y tienes 10% de descuento. Podemos expresar el problema directamente con Python.

````python
# Datos de entrada del problema
precio = 120
cantidad = 3
descuento = 0.10  # 10% expresado como decimal

# Operaciones intermedias
subtotal = precio * cantidad
total = subtotal * (1 - descuento)

# Salida: mostramos el resultado final
print(total)
````

### 💡 IDEA

Programar no empieza con sintaxis. Empieza con una pregunta:

**¿Qué información tengo, qué quiero obtener y qué pasos conectan ambas cosas?**

`datos → operaciones → resultado`

## 2. Python también puede funcionar como calculadora

Puedes ejecutar expresiones directamente.

````python
# Python puede evaluar una expresión y mostrar su resultado.
3 + 7
````

| Operador | Significado |
|---|---|
| `+` | suma |
| `-` | resta |
| `*` | multiplicación |
| `/` | división |
| `//` | división entera |
| `**` | potencia |
| `%` | residuo |

Los paréntesis ayudan a dejar clara la prioridad de las operaciones.

````python
# Los paréntesis ayudan a hacer explícito el orden de las operaciones.
resultado = (7 + 3 * (4 - 2**2) - 1) % 2
resultado  # en una notebook, la última expresión se muestra automáticamente
````

## 3. Variables: ponerle nombre a la información

Una variable permite guardar un valor y reutilizarlo después.

````python
# Creamos una variable llamada precio y guardamos el valor 120.
precio = 120

# Escribir el nombre permite inspeccionar su valor en una notebook.
precio
````

Los nombres descriptivos hacen que el código sea más fácil de entender.

```python
precio_unitario = 120
cantidad_productos = 3
```

es mucho más claro que:

```python
x = 120
y = 3
```

Para nombres compuestos usaremos normalmente `snake_case`.

## 4. Tipos de datos

Python distingue, entre otros:

- `int`: números enteros
- `float`: números con decimales
- `str`: texto
- `bool`: verdadero o falso

````python
# Cada variable guarda un tipo de dato distinto.
edad = 29                 # int: entero
temperatura = 22.5        # float: decimal
nombre = "Lucía"          # str: texto
activo = True             # bool: verdadero/falso

# type() nos dice qué tipo tiene cada objeto.
print(type(edad))
print(type(temperatura))
print(type(nombre))
print(type(activo))
````

El tipo determina qué operaciones tienen sentido. Sumar números no significa lo mismo que unir texto.

````python
# Con números, + suma.
print(10 + 5)

# Con strings, + concatena (une) texto.
print("Data" + " " + "Science")
````

## 5. Conversión de tipos

A veces recibimos información en un formato y necesitamos transformarla. Esto ocurre mucho con formularios, archivos o APIs.

````python
# Este valor se parece a un número, pero llega como texto.
edad_texto = "29"

# int() intenta convertirlo a un entero.
edad_numero = int(edad_texto)

print(edad_texto, type(edad_texto))
print(edad_numero, type(edad_numero))
````

### ⚠️ ERROR ÚTIL

¿Qué pasaría con `int("veintinueve")`?

Python no puede convertir cualquier texto en un número. Cuando algo falla, el error suele explicar qué intentó hacer y por qué no pudo.

## 6. Texto: trabajar con strings

Los strings sirven para representar texto. Las **f-strings** permiten insertar variables dentro de un mensaje.

````python
# Guardamos texto en variables.
nombre = "Lucía"
proyecto = "análisis de ventas"

# La f-string inserta valores dentro de un texto usando { }.
mensaje = f"{nombre} está trabajando en un {proyecto}."
print(mensaje)
````

## 7. Listas: varios valores bajo un mismo nombre

Cuando tenemos varios datos relacionados podemos agruparlos en una lista.

````python
# Una lista conserva varios valores en un orden definido.
ventas = [120, 180, 90, 250, 140]

print(ventas)      # lista completa
print(ventas[0])   # primer elemento: índice 0
print(ventas[-1])  # último elemento
print(len(ventas)) # cantidad de elementos
````

Los índices comienzan en `0`. También podemos seleccionar una parte de la lista con *slicing*.

````python
# El slicing [inicio:fin] extrae una parte de la lista.
# El límite final no se incluye; [:3] toma los índices 0, 1 y 2.
ventas[:3]
````

## 8. Diccionarios: información con significado

Una lista guarda valores en orden. Un diccionario guarda pares `clave: valor`, lo que permite describir una entidad de forma más clara.

````python
# Un diccionario guarda pares clave: valor.
producto = {
    "nombre": "Producto A",
    "precio": 120,
    "stock": 15
}

# Accedemos a un valor escribiendo su clave entre corchetes.
print(producto["nombre"])
print(producto["precio"])
````

Piensa en un diccionario como una pequeña ficha estructurada. Esta idea aparece constantemente en datos, APIs y configuraciones.

## 9. Un pequeño problema completo

Tenemos información de un producto y queremos calcular el valor total disponible en inventario.

````python
# Datos estructurados del producto.
producto = {
    "nombre": "Producto A",
    "precio": 120,
    "stock": 15
}

# Combinamos dos valores para obtener una nueva métrica.
valor_inventario = producto["precio"] * producto["stock"]

# Comunicamos el resultado con una f-string.
print(f"{producto['nombre']} tiene un valor de inventario de ${valor_inventario}.")
````

**información → operación → resultado → comunicación**

Esa misma lógica aparecerá después en análisis de datos, simulaciones y modelos.
