## 1. Programar es dividir un problema

Una computadora no “adivina” lo que queremos. Necesita instrucciones claras y en un orden específico.

Imagina una compra:

- cada producto cuesta 120 pesos;
- compras 3;
- tienes 10% de descuento.

Podemos expresar ese problema directamente con Python.

````python
precio = 120
cantidad = 3
descuento = 0.10

subtotal = precio * cantidad
total = subtotal * (1 - descuento)

print(total)
````

### 🧠 Idea Mau

Programar no empieza con sintaxis. Empieza con una pregunta:

**¿Qué información tengo, qué quiero obtener y qué pasos conectan ambas cosas?**

En el ejemplo anterior:

`datos → operaciones → resultado`

## 2. Python también puede funcionar como calculadora

Puedes ejecutar expresiones directamente.

````python
3 + 7
````

Operadores básicos:

| Operador | Significado |

## 3. Variables: ponerle nombre a la información

Una variable nos permite guardar un valor y reutilizarlo después.

En lugar de escribir `120` una y otra vez, podemos darle un nombre:

````python
precio = 120
precio
````

Los nombres descriptivos hacen que el código sea más fácil de entender.

```python
precio_unitario = 120
cantidad_productos = 3
```

es mucho más claro que:

```python
x = 120
y = 3
```

Para nombres compuestos usaremos normalmente `snake_case`.

## 4. Tipos de datos

Los valores no representan siempre lo mismo.

Python distingue, entre otros:

- `int`: números enteros
- `float`: números con decimales
- `str`: texto
- `bool`: verdadero o falso

````python
edad = 37
temperatura = 22.5
nombre = "Mau"
activo = True

print(type(edad))
print(type(temperatura))
print(type(nombre))
print(type(activo))
````

No necesitas memorizar los nombres todavía. Lo importante es reconocer que **el tipo determina qué operaciones tienen sentido**.

Por ejemplo, sumar números no significa lo mismo que unir texto.

````python
print(10 + 5)
print("Data" + " " + "Science")
````

## 5. Conversión de tipos

A veces recibimos información en un formato y necesitamos transformarla.

Esto ocurre mucho cuando los datos vienen de formularios, archivos o APIs.

````python
edad_texto = "37"
edad_numero = int(edad_texto)

print(edad_texto, type(edad_texto))
print(edad_numero, type(edad_numero))
````

### ⚠️ Error útil

Prueba mentalmente qué pasaría con:


## 6. Texto: trabajar con strings

Los strings sirven para representar texto.

````python
nombre = "Mauricio"
proyecto = "AELITYX"

mensaje = f"{nombre} está trabajando en {proyecto}."
print(mensaje)
````

Las **f-strings** permiten insertar variables dentro de texto y serán muy útiles para reportes, mensajes y resultados.

## 7. Listas: varios valores bajo un mismo nombre

Cuando tenemos varios datos relacionados podemos agruparlos en una lista.

````python
ventas = [120, 180, 90, 250, 140]

print(ventas)
print(ventas[0])
print(ventas[-1])
print(len(ventas))
````

Los índices comienzan en `0`.

Eso significa:

- `ventas[0]` → primer elemento
- `ventas[1]` → segundo elemento
- `ventas[-1]` → último elemento

También podemos seleccionar una parte de la lista.

````python
ventas[:3]
````

## 8. Diccionarios: información con significado

Una lista guarda valores en orden.  
Un diccionario guarda pares `clave: valor`.

Eso permite describir mejor una entidad.

````python
producto = {
    "nombre": "Producto A",
    "precio": 120,
    "stock": 15
}

print(producto["nombre"])
print(producto["precio"])
````

Piensa en un diccionario como una pequeña ficha estructurada.

Más adelante veremos que esta idea aparece constantemente en datos, APIs, configuraciones y sistemas reales.

## 9. Un pequeño problema completo

Vamos a juntar lo aprendido.

Tenemos información de un producto y queremos calcular el valor total disponible en inventario.

````python
producto = {
    "nombre": "Producto A",
    "precio": 120,
    "stock": 15
}

valor_inventario = producto["precio"] * producto["stock"]

print(f"{producto['nombre']} tiene un valor de inventario de ${valor_inventario}.")
````

Observa la estructura:

**información → operación → resultado → comunicación**

Esa misma lógica aparecerá después en análisis de datos, simulaciones y modelos.
