> **Objetivo:** representar, inspeccionar, limpiar, transformar y resumir datos tabulares.

Hasta ahora trabajamos principalmente con números. En la práctica solemos recibir **tablas**.

```python
# Importamos pandas y usamos el alias convencional pd.
import pandas as pd
```

## 1. Crear un DataFrame

```python
# Un diccionario de listas puede convertirse en una tabla.
datos = {
    "producto": ["A", "B", "C", "D", "E"],
    "categoria": ["Norte", "Norte", "Sur", "Sur", "Norte"],
    "ventas": [120, 180, 95, 210, 160],
    "unidades": [3, 4, 2, 5, 4]
}

# DataFrame es la estructura tabular principal de pandas.
df = pd.DataFrame(datos)
df
```

### 💡 IDEA

Un DataFrame organiza observaciones en filas y variables en columnas.

Una **Series** es esencialmente una columna con índice. Un **DataFrame** reúne varias Series alineadas en una tabla.

## 2. Inspeccionar antes de transformar

```python
# shape muestra (filas, columnas).
print("shape:", df.shape)

# columns contiene los nombres de las columnas.
print("columnas:", df.columns.tolist())

# display() es cómodo en notebooks porque renderiza la tabla visualmente.
display(df.head())  # head() muestra las primeras filas
```

```python
# info() resume tipos de datos y valores no nulos.
df.info()
```

Antes de limpiar o modelar, conviene saber **qué tenemos realmente**.

## 3. Seleccionar columnas y filas

```python
# Un solo nombre devuelve una Series.
df["ventas"]
```

```python
# Una lista de nombres devuelve un DataFrame con esas columnas.
df[["producto", "ventas"]]
```

## 4. Filtrar por una condición

```python
# Primero construimos una condición booleana y la usamos como filtro.
ventas_altas = df[df["ventas"] >= 150]
ventas_altas
```

## 5. Ordenar

```python
# Ordenamos de mayor a menor según la columna ventas.
df.sort_values("ventas", ascending=False)
```

## 6. Crear nuevas columnas

```python
# Creamos una columna calculada a partir de dos columnas existentes.
df["precio_promedio"] = df["ventas"] / df["unidades"]
df
```

La creación de columnas suele representar una hipótesis, una métrica o una transformación útil.

## 7. Valores faltantes

```python
# copy() evita modificar el DataFrame original.
df_faltantes = df.copy()

# .loc[fila, columna] permite seleccionar una posición por etiqueta.
df_faltantes.loc[2, "ventas"] = None

df_faltantes
```

```python
# isna() marca faltantes; sum() cuenta cuántos hay por columna.
print(df_faltantes.isna().sum())
```

No existe una única respuesta correcta para los valores faltantes. Podemos eliminarlos, imputarlos o conservarlos, pero la decisión debe tener sentido para el problema.

```python
# Ejemplo didáctico: rellenamos el faltante con la mediana de la columna.
# En un caso real, la estrategia depende del significado del dato.
df_faltantes["ventas"] = df_faltantes["ventas"].fillna(
    df_faltantes["ventas"].median()
)
df_faltantes
```

## 8. Agrupar y resumir

```python
# Agrupamos por categoria y calculamos varias métricas por grupo.
resumen = (
    df.groupby("categoria", as_index=False)
      .agg(
          ventas_totales=("ventas", "sum"),
          venta_promedio=("ventas", "mean"),
          unidades_totales=("unidades", "sum")
      )
)

resumen
```

### 💡 IDEA

`groupby` responde preguntas del tipo:

**“¿Cómo cambia esta métrica entre grupos?”**

## 10. Un flujo de análisis pequeño

```python
df_analisis = df.copy()

# Encadenamos transformaciones para leer el flujo de arriba hacia abajo.
resultado = (
    df_analisis
    .assign(ticket=df_analisis["ventas"] / df_analisis["unidades"])
    .query("ventas >= 100")
    .sort_values("ventas", ascending=False)
)

resultado
```
