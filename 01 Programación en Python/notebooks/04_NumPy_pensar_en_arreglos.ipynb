```python
# Importamos NumPy y usamos el alias convencional np.
import numpy as np
```

## 1. De una lista a un arreglo

```python
# Lista normal de Python.
lista = [10, 20, 30, 40]

# np.array convierte la lista en un arreglo NumPy.
arreglo = np.array(lista)

print(lista)
print(arreglo)
print(type(arreglo))
```

Con una lista, multiplicar por 2 repite elementos. Con un arreglo de NumPy, multiplica cada valor.

```python
# En una lista, * 2 repite los elementos.
print(lista * 2)

# En NumPy, * 2 opera elemento por elemento.
print(arreglo * 2)
```

### 💡 IDEA

La vectorización permite expresar operaciones sobre colecciones completas sin escribir un ciclo explícito para cada elemento.

## 2. Forma, dimensiones y tipo

```python
# Un arreglo puede tener más de una dimensión.
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("shape:", matriz.shape)  # filas, columnas
print("ndim:", matriz.ndim)    # número de dimensiones
print("size:", matriz.size)    # total de elementos
print("dtype:", matriz.dtype)  # tipo interno de los datos
```

`shape` responde una pregunta muy importante: **¿cómo están organizados los datos?**

## 3. Crear arreglos útiles

```python
# Funciones útiles para construir arreglos sin escribir todos los valores.
print(np.zeros(5))          # cinco ceros
print(np.ones(3))           # tres unos
print(np.arange(0, 10, 2))  # 0, 2, 4, 6, 8
print(np.linspace(0, 1, 5)) # cinco puntos equidistantes entre 0 y 1
```

## 4. Indexación y slicing

```python
valores = np.array([5, 10, 15, 20, 25, 30])

print(valores[0])    # primer elemento
print(valores[-1])   # último elemento
print(valores[1:4])  # índices 1, 2 y 3
```

```python
# Creamos 12 valores y los reorganizamos como matriz 3 x 4.
matriz = np.arange(1, 13).reshape(3, 4)

print(matriz)
print("Primera fila:", matriz[0])
print("Segunda columna:", matriz[:, 1])  # : significa todas las filas
```

## 5. Operaciones vectorizadas

### Broadcasting, en una frase

NumPy puede combinar arreglos y escalares —y algunos arreglos de formas compatibles— sin que tú escribas ciclos manuales. A esta regla de expansión se le llama **broadcasting**.

Por ahora quédate con la intuición: `arreglo * 2` significa “multiplica cada elemento por 2”.

```python
precios = np.array([100, 120, 80, 150])
descuento = 0.10

# El escalar (1 - descuento) se aplica a cada elemento del arreglo.
precios_finales = precios * (1 - descuento)
print(precios_finales)
```

## 6. Resumir datos

```python
mediciones = np.array([12.4, 15.1, 13.8, 16.2, 14.5])

# Métodos de agregación: convierten muchos valores en un resumen.
print("promedio:", mediciones.mean())
print("mínimo:", mediciones.min())
print("máximo:", mediciones.max())
print("desviación estándar:", mediciones.std())
```

## 7. Máscaras booleanas: filtrar con una condición

```python
ventas = np.array([75, 180, 90, 250, 320, 110])

# La comparación produce un arreglo de True/False.
mascara = ventas >= 150

print(mascara)
print(ventas[mascara])  # conservamos sólo posiciones True
```

## 8. Cambiar la forma

```python
datos = np.arange(12)

# reshape cambia la organización, no los valores.
matriz = datos.reshape(3, 4)
print(matriz)
```

Cambiar la forma no inventa datos. Sólo cambia cómo los organizamos.

## 9. Números aleatorios y reproducibilidad

```python
# Creamos un generador pseudoaleatorio reproducible.
rng = np.random.default_rng(42)

# 10 muestras de una normal con media 100 y desviación 15.
muestras = rng.normal(loc=100, scale=15, size=10)
print(muestras)
```

La semilla `42` hace reproducible la secuencia. Eso importa cuando queremos comparar experimentos.

## 10. Primer contacto con simulación

```python
# Reiniciamos el generador con la misma semilla para reproducibilidad.
rng = np.random.default_rng(42)

# Generamos 1 000 escenarios posibles.
ventas_simuladas = rng.normal(loc=1000, scale=180, size=1000)

# Resumimos el conjunto de escenarios.
print("Escenarios:", ventas_simuladas.size)
print("Promedio simulado:", ventas_simuladas.mean())
print("P5:", np.percentile(ventas_simuladas, 5))
print("P95:", np.percentile(ventas_simuladas, 95))
```

### 💡 IDEA

Acabamos de generar **1 000 escenarios posibles** y resumir su distribución.

Más adelante esta misma lógica será la base de simulaciones Monte Carlo.
