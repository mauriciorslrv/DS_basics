# 03 · Código reutilizable

> **Objetivo:** organizar lógica en funciones y módulos, trabajar con archivos, interpretar errores y entender la idea esencial de objetos.

No necesitas convertir cada programa en una arquitectura compleja. La meta es reconocer cuándo conviene **separar responsabilidades**.

## 1. De una celda larga a una función clara

```python
def convertir_celsius_a_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

temperaturas = [18, 22, 30]
convertidas = [convertir_celsius_a_fahrenheit(t) for t in temperaturas]

print(convertidas)
```

### 💡 IDEA

Si una parte del código tiene nombre, propósito y entradas claras, probablemente puede convertirse en una función.

## 2. Módulos: código que vive en otro archivo

Python permite guardar funciones en archivos `.py` e importarlas después.

```python
from pathlib import Path

contenido_modulo = '''
def promedio(valores):
    return sum(valores) / len(valores)

def rango(valores):
    return max(valores) - min(valores)
'''.strip()

Path("utilidades_demo.py").write_text(contenido_modulo, encoding="utf-8")
print("Módulo creado.")
```

```python
import importlib
import utilidades_demo

importlib.reload(utilidades_demo)

datos = [10, 12, 15, 20]
print(utilidades_demo.promedio(datos))
print(utilidades_demo.rango(datos))
```

También puedes importar una definición concreta:

```python
from math import sqrt

print(sqrt(81))
```

Evita acostumbrarte a `from modulo import *`: hace menos claro de dónde viene cada nombre y puede provocar colisiones.

## 3. Archivos: leer y escribir información

Los programas reales suelen interactuar con archivos.

```python
from pathlib import Path

ruta = Path("mediciones.txt")
ruta.write_text("12.4\n15.1\n13.8\n", encoding="utf-8")

contenido = ruta.read_text(encoding="utf-8")
print(contenido)
```

Podemos convertir ese texto en números.

```python
valores = [float(linea) for linea in contenido.splitlines()]
print(valores)
print(sum(valores) / len(valores))
```

## 4. Errores: información para depurar

Los errores no son un fracaso del proceso. Son una descripción de qué no pudo hacer Python.

```python
texto = "42"

try:
    numero = int(texto)
    print(numero)
except ValueError:
    print("El texto no se puede convertir a entero.")
```

### ⚠️ ERROR ÚTIL

Usa `try/except` cuando esperas una falla razonable y sabes cómo responder a ella. No lo uses para esconder errores que todavía no entiendes.

## 5. Objetos: estado + comportamiento

Muchos elementos de Python son objetos. Una lista, un DataFrame y un modelo de machine learning tienen datos internos y métodos que podemos llamar.

Una clase define ese tipo de objeto.

```python
class Experimento:
    def __init__(self, nombre, resultados):
        self.nombre = nombre
        self.resultados = resultados

    def promedio(self):
        return sum(self.resultados) / len(self.resultados)

experimento = Experimento("Prueba A", [12, 15, 18, 21])

print(experimento.nombre)
print(experimento.promedio())
```

### 💡 IDEA

No necesitas dominar programación orientada a objetos para analizar datos.

Por ahora basta con reconocer:

- **atributos**: información asociada al objeto;
- **métodos**: acciones que el objeto sabe realizar.

Cuando más adelante veas `df.head()` o `modelo.fit()`, esta idea ya no será extraña.

## 6. Un pequeño flujo organizado

```python
def cargar_valores(ruta):
    texto = Path(ruta).read_text(encoding="utf-8")
    return [float(linea) for linea in texto.splitlines()]

def resumir(valores):
    return {
        "cantidad": len(valores),
        "minimo": min(valores),
        "maximo": max(valores),
        "promedio": sum(valores) / len(valores)
    }

datos = cargar_valores("mediciones.txt")
resumen = resumir(datos)
print(resumen)
```
