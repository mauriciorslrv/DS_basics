## 1. Una pregunta antes del código

Supón que registramos una temperatura y queremos clasificarla como baja, normal o alta.

La lógica humana sería:

`dato → comparación → decisión → resultado`

```python
# Dato que queremos clasificar.
temperatura = 31

# Python revisa las condiciones de arriba hacia abajo.
if temperatura < 15:
    estado = "baja"
elif temperatura <= 28:
    estado = "normal"
else:
    estado = "alta"

print(estado)
```

### 💡 IDEA

Un `if` no “piensa”. Evalúa una condición que produce `True` o `False` y sigue una ruta.

### La indentación también es sintaxis

En Python, los espacios al inicio de una línea indican qué instrucciones pertenecen a un `if`, un `for`, un `while` o una función.

Usaremos normalmente **4 espacios** por nivel. Si la indentación está mal, Python no puede interpretar correctamente la estructura del programa.

## 2. Comparaciones y condiciones

Las comparaciones son preguntas que Python puede evaluar.

```python
# Guardamos un valor para comparar.
valor = 12

# Cada comparación produce True o False.
print(valor > 10)       # mayor que
print(valor == 12)      # igual a
print(valor != 5)       # distinto de
print(10 <= valor <= 20)  # comparación encadenada
```

También podemos combinar condiciones con `and`, `or` y `not`.

```python
edad = 24
tiene_registro = True

# and exige que ambas condiciones sean verdaderas.
puede_ingresar = edad >= 18 and tiene_registro
print(puede_ingresar)
```

## 3. Repetir una tarea con `for`

Cuando una operación debe aplicarse a varios elementos, repetir el código manualmente no escala.

```python
# Recorremos cada elemento de la lista una vez.
mediciones = [18.2, 19.1, 21.4, 20.0]

for medicion in mediciones:
    print(f"Medición: {medicion}")
```

Podemos combinar un ciclo con una decisión.

```python
ventas = [120, 75, 310, 95, 220]

# El ciclo visita una venta y el if decide cómo clasificarla.
for venta in ventas:
    if venta >= 200:
        print(venta, "→ venta alta")
    else:
        print(venta, "→ venta regular")
```

### Una forma compacta: comprensiones de lista

Cuando queremos **crear una nueva lista aplicando una transformación sencilla**, Python ofrece una sintaxis compacta llamada *list comprehension*.

Primero piensa en el `for` normal; después reconoce su versión compacta.

```python
# Versión explícita con for.
cuadrados = []
for numero in [1, 2, 3, 4]:
    cuadrados.append(numero ** 2)

print(cuadrados)

# La misma idea en una sola expresión.
cuadrados_compactos = [numero ** 2 for numero in [1, 2, 3, 4]]
print(cuadrados_compactos)
```

## 4. `range()` cuando importa el número de repeticiones

```python
# range(1, 6) genera 1, 2, 3, 4 y 5.
for intento in range(1, 6):
    print(f"Intento {intento}")
```

## 5. `while` cuando importa una condición

`while` repite una tarea mientras una condición siga siendo verdadera.

```python
energia = 3

# El bloque se repite mientras energia sea mayor que cero.
while energia > 0:
    print("Procesando...")
    energia -= 1  # equivale a: energia = energia - 1

print("Proceso terminado")
```

## 6. Funciones: guardar una idea reutilizable

Una función agrupa una tarea bajo un nombre.

```python
# def crea una función; precio y cantidad son parámetros.
def calcular_total(precio, cantidad):
    # return devuelve el resultado al lugar donde se llamó la función.
    return precio * cantidad

print(calcular_total(120, 3))
print(calcular_total(85, 5))
```

Una buena función suele hacer **una cosa concreta** y tener un nombre que describa su propósito.

## 7. Parámetros y resultados

```python
def clasificar_valor(valor, umbral=100):
    # umbral=100 es un valor por defecto.
    if valor >= umbral:
        return "alto"
    return "bajo"

print(clasificar_valor(140))
print(clasificar_valor(80))
print(clasificar_valor(80, umbral=50))  # aquí reemplazamos el valor por defecto
```

## 8. Tuplas y conjuntos: sólo cuando tienen sentido

Una **tupla** es útil para agrupar valores que no planeamos modificar.

```python
# Una tupla usa paréntesis y mantiene un orden fijo de valores.
coordenada = (19.43, -99.13)
print(coordenada[0], coordenada[1])
```

Un **set** es útil cuando queremos valores únicos.

```python
categorias = ["A", "B", "A", "C", "B"]

# set(...) elimina duplicados porque un conjunto guarda valores únicos.
categorias_unicas = set(categorias)
print(categorias_unicas)
```

## 9. Un problema completo

Queremos revisar una lista de observaciones, conservar sólo las válidas y calcular su promedio.

```python
observaciones = [12, -3, 18, 21, -1, 15]
validas = []  # aquí iremos guardando sólo los valores aceptados

for valor in observaciones:
    if valor >= 0:
        validas.append(valor)  # append agrega un elemento al final de la lista

# Calculamos el promedio únicamente con los valores válidos.
promedio = sum(validas) / len(validas)

print("Observaciones válidas:", validas)
print("Promedio:", promedio)
```

### 💡 IDEA

Aquí aparecen tres piezas fundamentales:

**representar datos → aplicar reglas → resumir resultados**
