> **Objetivo:** elegir gráficos según la pregunta y usarlos para explorar, comparar y comunicar información.

Una gráfica no es decoración. Es una herramienta para responder una pregunta.

```python
# Matplotlib crea visualizaciones; pandas y NumPy ayudan a preparar los datos.
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## 1. La pregunta va primero

Antes de graficar, decide qué quieres entender:

- evolución en el tiempo;
- comparación entre categorías;
- distribución de valores;
- relación entre variables;
- posibles valores atípicos.

### 💡 IDEA

**Una gráfica bonita que no responde ninguna pregunta sigue siendo una mala gráfica.**

## 2. Línea: evolución en el tiempo

```python
dias = ["Lun", "Mar", "Mié", "Jue", "Vie"]
visitas = [120, 135, 128, 160, 175]

# figure() define el tamaño del lienzo.
plt.figure(figsize=(8, 4.5))

# plot() dibuja una línea; marker agrega un punto en cada observación.
plt.plot(dias, visitas, marker="o")
plt.title("Visitas por día")
plt.xlabel("Día")
plt.ylabel("Visitas")
plt.show()  # renderiza la figura
```

## 3. Barras: comparar categorías

```python
categorias = ["A", "B", "C", "D"]
ventas = [320, 250, 410, 180]

plt.figure(figsize=(8, 4.5))
plt.bar(categorias, ventas)  # barras para comparar categorías
plt.title("Ventas por categoría")
plt.xlabel("Categoría")
plt.ylabel("Ventas")
plt.show()
```

## 4. Histograma: entender una distribución

```python
# Datos simulados para observar una distribución.
rng = np.random.default_rng(42)
tiempos = rng.normal(loc=30, scale=6, size=300)

plt.figure(figsize=(8, 4.5))
plt.hist(tiempos, bins=20)  # bins controla cuántos intervalos usamos
plt.title("Distribución de tiempos")
plt.xlabel("Tiempo")
plt.ylabel("Frecuencia")
plt.show()
```

Un histograma permite observar concentración, dispersión, asimetrías y valores poco frecuentes.

## 5. Dispersión: relación entre dos variables

```python
inversion = np.array([10, 15, 20, 25, 30, 35, 40])
resultados = np.array([12, 18, 22, 27, 33, 35, 43])

plt.figure(figsize=(8, 4.5))
plt.scatter(inversion, resultados)  # un punto por cada par (x, y)
plt.title("Inversión y resultado")
plt.xlabel("Inversión")
plt.ylabel("Resultado")
plt.show()
```

### ⚠️ ERROR ÚTIL

Una relación visual no demuestra causalidad. La gráfica puede sugerir una pregunta; no sustituye el análisis.

## 6. Boxplot: distribución y posibles atípicos

```python
grupo_a = [12, 13, 15, 14, 16, 15, 17, 40]
grupo_b = [18, 19, 17, 20, 21, 19, 18, 22]

plt.figure(figsize=(8, 4.5))
# El boxplot resume mediana, cuartiles y posibles valores atípicos.
plt.boxplot([grupo_a, grupo_b], tick_labels=["Grupo A", "Grupo B"])
plt.title("Comparación de distribuciones")
plt.ylabel("Valor")
plt.show()
```

## 7. Una tabla, varias preguntas

```python
# Creamos una tabla pequeña para formular varias preguntas.
df = pd.DataFrame({
    "mes": ["Ene", "Feb", "Mar", "Abr", "May", "Jun"],
    "ventas": [120, 140, 135, 170, 190, 185],
    "costos": [80, 85, 90, 105, 115, 118]
})

# Nueva métrica: diferencia entre ventas y costos.
df["margen"] = df["ventas"] - df["costos"]
df
```

```python
plt.figure(figsize=(8, 4.5))

# Dos series en el mismo gráfico para comparar su evolución.
plt.plot(df["mes"], df["ventas"], marker="o", label="Ventas")
plt.plot(df["mes"], df["costos"], marker="o", label="Costos")

plt.title("Ventas y costos por mes")
plt.xlabel("Mes")
plt.ylabel("Valor")
plt.legend()  # muestra las etiquetas de cada serie
plt.show()
```

## 8. Diseñar para leer, no para impresionar

Buenas prácticas mínimas:

- título que diga qué estamos viendo;
- ejes con nombres comprensibles;
- escala adecuada;
- evitar elementos que no aportan;
- no saturar de categorías;
- destacar sólo cuando haya una razón.

## 9. Del gráfico a la interpretación

Una visualización no termina al mostrarla. Hay que traducirla a una observación.

Por ejemplo:

> “Las ventas crecen de enero a mayo, pero en junio se estabilizan mientras los costos siguen aumentando.”

Esa frase genera una nueva pregunta: **¿el margen también se está frenando?**

```python
plt.figure(figsize=(8, 4.5))

# Ahora graficamos la métrica que respondió la nueva pregunta: el margen.
plt.bar(df["mes"], df["margen"])
plt.title("Margen por mes")
plt.xlabel("Mes")
plt.ylabel("Margen")
plt.show()
```

### 💡 IDEA

El ciclo útil es:

**pregunta → gráfico → patrón → interpretación → nueva pregunta**
