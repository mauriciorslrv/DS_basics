## 🎯 ¿Qué es un Perceptrón?

Un **Perceptrón** es un algoritmo de clasificación binaria que:

1. **Recibe entradas** (características numéricas)
2. **Calcula una suma ponderada** con pesos aprendidos
3. **Aplica una función de activación** (escalón)
4. **Produce una salida binaria** (0 o 1, -1 o 1)

### 🔄 Funcionamiento básico

```
Entrada → Pesos → Suma → Activación → Salida
   x₁       w₁      Σ      f(z)       ŷ
   x₂       w₂             z = w·x + b
   ...      ...
   xₙ       wₙ
```

### 📈 Regla de Aprendizaje

El perceptrón aprende ajustando sus pesos usando:

$$w_{nuevo} = w_{anterior} + \eta \cdot (y_{real} - y_{predicho}) \cdot x$$

donde:
- $\eta$: tasa de aprendizaje
- $y_{real}$: etiqueta verdadera
- $y_{predicho}$: predicción del modelo
- $x$: vector de características

# 📖 Fundamentos Teóricos del Perceptrón

---
El Perceptron simple, también conocido como **red neuronal de una sola capa (Single-Layer Neural Network)**, es un algoritmo de clasificación binaria creado por **Frank Rosenblatt** en 1958.

---

## 🧠 Inspiración Biológica

La neurona recibe **impulsos externos** (considerados como entradas - inputs) con distinta importancia (o **pesos**) en una **función de activación**. Si el estímulo agregado sobrepasa cierto umbral, la neurona se activa.

### 📊 Matemáticas del Perceptrón

Definimos:
- **x** como el vector de estímulos: $x = [x_1, x_2, ..., x_m]$
- **w** como el vector de pesos: $w = [w_1, w_2, ..., w_m]$
- **z** como la función de activación lineal: $z = w^T x + b$
- **φ(z)** como la función de activación final

**Ecuación completa:**
$$z = w_1 \cdot x_1 + w_2 \cdot x_2 + \dots + w_m \cdot x_m + b$$

$$z = \sum_{i=1}^{m} w_i \cdot x_i + b = w^T x + b$$

---

## ⚡ Función de Activación: Escalón de Heaviside

El perceptrón φ(z) se considera **activo** cuando su valor es mayor o igual al umbral θ, o **inactivo** en cualquier otro caso.

**Formalmente:**
$$\phi(z) = \begin{cases}
1 & \text{si } z \geq \theta \\
-1 & \text{si } z < \theta
\end{cases}$$

Esta es la **función escalón** (step function) o **función de Heaviside**.

### 💡 Nota Importante
En la práctica, usualmente se usa **θ = 0** y se absorbe el umbral en el bias **b = -θ**, simplificando a:
$$\phi(z) = \begin{cases}
1 & \text{si } z \geq 0 \\
-1 & \text{si } z < 0
\end{cases}$$

---

## 🎓 Regla de Aprendizaje

El perceptrón tiene una regla de aprendizaje bastante simple que le permite ir ajustando los valores de los pesos (w). Para ello, se siguen los siguientes pasos:

### 🔄 Proceso de Entrenamiento

1. **Inicialización**: Asignar valores iniciales a los pesos (0 o valores pequeños aleatorios)
2. **Para cada muestra de entrenamiento** x⁽i⁾ hacer:
   - Computar el valor de salida ŷ⁽i⁾
   - Actualizar los pesos usando la regla de aprendizaje

### 📈 Actualización de Pesos

La actualización de los pesos se hace incrementando o disminuyendo Δwⱼ:

$$w_j = w_j + \Delta w_j$$

$$\Delta w_j = \eta \cdot (y^{(i)} - \hat{y}^{(i)}) \cdot x_j^{(i)}$$

### 🔍 Interpretación de la Fórmula

Donde:
- **η** (eta): tasa de aprendizaje (valor entre 0 y 1.0)
- **y⁽i⁾**: valor real (etiqueta verdadera)
- **ŷ⁽i⁾**: valor de salida calculado (predicción)
- **xⱼ⁽i⁾**: valor de la muestra asociado al peso wⱼ

### 💭 ¿Qué significa esto?

Esto implica que **si el valor real y el valor calculado son iguales**, w no se actualiza (Δwⱼ = 0).

Sin embargo, **si hubo un error en la predicción**, el valor será actualizado en la diferencia entre el valor real y el predicho, ajustado por el valor de la muestra y la tasa de aprendizaje.

---

## 🎯 Resumen Conceptual

El perceptrón aprende una **frontera de decisión lineal** que separa las dos clases en el espacio de características. Es el algoritmo de ML más simple pero poderoso para problemas linealmente separables.

## 🧠 La Regla de Aprendizaje del Perceptrón

El perceptrón utiliza una **regla de aprendizaje** simple pero poderosa que le permite ajustar automáticamente los pesos para mejorar sus predicciones. Este es el corazón del aprendizaje automático supervisado.

### 📋 Algoritmo de Entrenamiento:

1. **Inicialización**: Asignar valores iniciales a los pesos (generalmente 0 o valores pequeños aleatorios)
2. **Para cada muestra de entrenamiento** x^(i):
   - Calcular la salida predicha ŷ^(i)
   - Actualizar los pesos según el error

### 🔄 Actualización de Pesos:

La actualización se realiza incrementando o disminuyendo cada peso w_j en Δw_j:

$$ w_j = w_j + \Delta w_j $$

$$ \Delta w_j = \eta (y^{(i)} - \hat{y}^{(i)}) x_j^{(i)} $$

Donde:
- **η** (eta): Tasa de aprendizaje (entre 0 y 1)
- **y^(i)**: Valor real (etiqueta correcta)
- **ŷ^(i)**: Valor predicho por el perceptrón
- **x_j^(i)**: Valor de la j-ésima característica de la muestra i

### 💡 Interpretación:
- Si la predicción es correcta (y = ŷ), Δw_j = 0 → no hay cambio
- Si la predicción es incorrecta, los pesos se ajustan proporcionalmente al error, la tasa de aprendizaje y el valor de la entrada

Esta regla garantiza que el perceptrón converja hacia una solución óptima para problemas linealmente separables.

# 💻 Implementación del Perceptrón en Python

## 🏗️ Estructura de la Clase

Implementaremos una clase Perceptrón que tenga los siguientes métodos:

* **`__init__`**: Constructor que define la tasa de aprendizaje y número de pasadas
* **`fit`**: Implementa la regla de aprendizaje, inicializa pesos y los ajusta
* **`predict`**: Función escalón φ(z) - retorna 1 si z ≥ 0, -1 si z < 0
* **`net_input`**: Implementación de z = w·x + b (función de activación lineal)

### 📝 Parámetros de la Clase

```python
class Perceptron:
    def __init__(self, eta=0.1, n_iter=10):
        # eta: tasa de aprendizaje (learning rate)
        # n_iter: número de pasadas sobre el dataset
```

### 🎯 Atributos Aprendidos

```python
self.w_: array-1d  # Pesos finales después del ajuste
self.errors_: list  # Número de errores en cada pasada
```

---

## 🔧 Implementación Paso a Paso

````python
# ============================================
# CLASE PERCEPTRÓN - IMPLEMENTACIÓN DESDE CERO
# ============================================

import numpy as np

class Perceptron:
    """
    Clasificador Perceptrón basado en el libro "Python Machine Learning" de Sebastian Raschka.

    Esta implementación es pedagógica - muestra cómo funciona el algoritmo internamente.

    Parámetros:
    -----------
    eta : float
        Tasa de aprendizaje (learning rate) - controla velocidad de aprendizaje
        Valores típicos: 0.01, 0.1, 0.5
    n_iter : int
        Número de pasadas sobre el dataset (épocas/epochs)
        Más épocas = más aprendizaje pero riesgo de overfitting

    Atributos:
    ----------
    w_ : array-1d, shape = [n_features + 1]
        Pesos finales después del ajuste (incluye bias en w_[0])
    errors_ : list
        Número de errores de clasificación en cada época
    """

    def __init__(self, eta=0.1, n_iter=10):
        """
        Constructor del Perceptrón

        Args:
            eta (float): Tasa de aprendizaje η
            n_iter (int): Número de iteraciones sobre el dataset
        """
        self.eta = eta      # Learning rate
        self.n_iter = n_iter  # Número de épocas

    def fit(self, X, y):
        """
        Ajustar el modelo a los datos de entrenamiento usando la regla del perceptrón.

        Parámetros:
        -----------
        X : array-like, shape = [n_samples, n_features]
            Vectores de entrenamiento
            n_samples: número de muestras
            n_features: número de características por muestra

        y : array-like, shape = [n_samples]
            Valores objetivo (etiquetas): deben ser {-1, 1}

        Retorna:
        --------
        self : object
            Retorna la instancia del perceptrón entrenado
        """
        # ============================================
        # INICIALIZACIÓN DE PESOS
        # ============================================
        # Creamos un vector de pesos con n_features + 1 elementos
        # w_[0] será el bias (sesgo), w_[1:] serán los pesos de las características
        self.w_ = np.zeros(1 + X.shape[1])

        # Lista para almacenar errores por época
        self.errors_ = []

        # ============================================
        # BUCLE DE ENTRENAMIENTO
        # ============================================
        for epoch in range(self.n_iter):
            errors = 0  # Contador de errores en esta época

            # Iterar sobre cada muestra de entrenamiento
            for xi, target in zip(X, y):
                # Calcular actualización usando regla del perceptrón
                update = self.eta * (target - self.predict(xi))

                # Actualizar pesos: w = w + η * error * x
                # w_[1:] son pesos de características, w_[0] es bias
                self.w_[1:] += update * xi  # Actualizar pesos de características
                self.w_[0] += update        # Actualizar bias

                # Contar errores (update != 0 significa predicción incorrecta)
                errors += int(update != 0.0)

            # Guardar número de errores de esta época
            self.errors_.append(errors)

        return self

    def predict(self, X):
        """
        Retornar clase usando función escalón de Heaviside.

        φ(z) = 1 si z ≥ θ; -1 en otro caso
        (Aquí usamos θ = 0)

        Parámetros:
        -----------
        X : array-like, shape = [n_samples, n_features]
            Muestras a clasificar

        Retorna:
        --------
        array-like, shape = [n_samples]
            Predicciones de clase: -1 o 1
        """
        # Calcular entrada neta z = w·x + b
        z = self.net_input(X)

        # Aplicar función escalón
        phi = np.where(z >= 0.0, 1, -1)

        return phi

    def net_input(self, X):
        """
        Calcular el valor z (net input) = w·x + b

        Esta es la suma ponderada de entradas más el bias.

        Parámetros:
        -----------
        X : array-like, shape = [n_samples, n_features]

        Retorna:
        --------
        array-like, shape = [n_samples]
            Valores z = w·x + b para cada muestra
        """
        # z = suma(w_i * x_i) + w_0 (bias)
        # w_[1:] son pesos de características, w_[0] es bias
        z = np.dot(X, self.w_[1:]) + self.w_[0]

        return z

# 💡 Esta implementación es básica pero funcional
# En la Parte 2 exploraremos versiones más avanzadas con funciones de activación sigmoide
````

## ¿Cómo hacer un perceptrón?

La estructura básica de un perceptrón es la siguiente:

    Inputs: Datos con los cuales quieres clasificar.
    Weights: Constantes que multiplican a incógnitas (inputs) de la ecuación.
    Bias: Constante que permite que se tome una decisión.
    Threshold: Punto que representa la división de las clasificaciones.
    Output: Predicción de la clasificación.
    
---

### Explicación general del proceso

Los inputs son los valores con los que esperas obtener una respuesta (output). Cada input es multiplicado por el correspondiente weight (los cuales inician con valor aleatorio), luego los resultados son sumados junto con una constante (bias). Si este resultado supera el threshold determina un output x, de lo contrario determina un output y.

# ⚙️ Reglas de Ajuste de Pesos

## 🎯 ¿Cuándo ajustar los pesos?

Cuando la **predicción es incorrecta**, debemos ajustar los pesos para corregir el error.

### 📊 Reglas de Decisión

- Si **z ≥ 0** → Predicción = 1
- Si **z < 0** → Predicción = -1

---

## 🔧 Fórmulas de Actualización

Las fórmulas para ajustar los pesos son:

### Para el Bias (w₀):
$$w_0 = w_0 + \eta \cdot (y_{real} - y_{predicho})$$

### Para los Pesos de Características (wⱼ):
$$w_j = w_j + \eta \cdot (y_{real} - y_{predicho}) \cdot x_j$$

### 🔍 Interpretación:

- **η**: Tasa de aprendizaje (learning rate)
- **y_real**: Clase verdadera {-1, 1}
- **y_predicho**: Clase predicha {-1, 1}
- **x_j**: Valor de la característica j

### 💭 ¿Por qué funciona?

- Si **error = 0** → No hay ajuste (predicción correcta)
- Si **error ≠ 0** → Ajuste proporcional al error y a las características
- La dirección del ajuste depende del signo del error

---

## 🎨 Representación Visual

La línea de decisión final tendrá la ecuación:
$$w_0 + w_1 \cdot x_1 + w_2 \cdot x_2 = 0$$

Que podemos reescribir como:
$$x_2 = -\frac{w_1}{w_2} \cdot x_1 - \frac{w_0}{w_2}$$

### Pero **¿qué representan estos weights?** 

Pues son las constantes que multiplican a incógnitas (inputs) de la ecuación, sólo recuerda que el $w_0$ multiplica a un 1, es decir, es una constante. Podemos hacer la ecuación de una recta con estos weights, de la siguiente forma:

$- 0.97 + 0.66x + 0.8y$

En la cual podemos despejar la y para tener una ecuación más bonita:

$y = (0.97 - 0.66x) / 0.8$

$y = 1.2125 - 0.825 x$

Ahora para hacer todo más claro y entendible haré la clase Perceptrón que tenga los siguientes métodos:

* generate_random_weights
* train
* predict
* adjust_weights
* verify

El constructor de la clase Perceptrón recibirá el número de weights y un array de las dos clases que se mapean a -1 y 1.

````python
class Perceptron:
    def __init__(self, number_of_weights, classes):
        self.number_of_weights = number_of_weights
        self.weights = self.generate_random_weights(number_of_weights)
        self.dict_classes = { classes[0]:1, classes[1]:-1 }
        
    def generate_random_weights(self, n):
        weights = []
        for x in range(n):
            weights.append(random.random()*10-5)
        return weights

    def predict(self, datum):
        weights_without_bias = self.weights[1:self.number_of_weights]
        attribute_values = datum[:self.number_of_weights-1]
        weight_bias = self.weights[0]
        activation = sum([i*j for i,j in zip(weights_without_bias,attribute_values)]) + weight_bias
        return 1 if activation > 0 else -1
    
    def adjust_weights(self, real_class, datum):
        self.weights[0] = self.weights[0] + real_class
        for i in range(1,self.number_of_weights):
            self.weights[i] = self.weights[i] + real_class * datum[i-1]
            
    def train(self, data, epochs):
        for epoch in range(epochs):
            print('Epoch {}'.format(epoch))
            for datum in data:
                real_class = self.dict_classes[datum[len(datum)-1]]
                prediction_class = self.predict(datum)
                if real_class != prediction_class:
                    self.adjust_weights(real_class,datum)
                    print('Adjusted weights: {}'.format(self.weights))
            print('Final weights from epoch {}: {}'.format(epoch,self.weights))
            
    def verify(self, data):
        count = 0
        for datum in data:
            real_class = self.dict_classes[datum[len(datum)-1]]
            prediction_class = self.predict(datum)
            if real_class != prediction_class:
                count = count + 1
        return (1-count/len(data))*100
````

La generación de weights aleatorios es muy sencilla, sólo intenta que abarquen números negativos y positivos, yo lo hice con un rango de -5 a 5 (puedes usar el rango que tu prefieras)

````python
def generate_random_weights(self, n):
    weights = []
    for x in range(n):
        weights.append(random.random()*10-5)
    return weights
````

La predicción es la suma de las multiplicaciones de los atributos por los weights más el bias (el bias lo deje como el primer elemento del array weights). Si el resultado es positivo predecimos con un 1, de lo contrario predecimos con un -1.

````python
def predict(self, datum):
    weights_without_bias = self.weights[1:self.number_of_weights]
    attribute_values = datum[:self.number_of_weights-1]
    weight_bias = self.weights[0]
    activation = sum([i*j for i,j in zip(weights_without_bias,attribute_values)]) + weight_bias
    return 1 if activation > 0 else -1
````

Ahora ¿Qué pasa si nuestra predicción es incorrecta?, tenemos que ajustar los weights, te recuerdo las fórmulas:

$$w_0 = w_0 + clase * bias$$

$$w_n = w_n + clase * x_n$$

---

La w es el weight, la clase es -1 o 1, x es el atributo y el bias siempre es 1. Con clase nos referimos a la clase real, no a la predicción.

````python
def adjust_weights(self, real_class, datum):
    self.weights[0] = self.weights[0] + real_class
    for i in range(1,self.number_of_weights):
        self.weights[i] = self.weights[i] + real_class * datum[i-1]
````

Puedes notar que el único diferente es el bias, que lo que hacemos es simplemente sumarle la clase real. Los demás weights se calculan al sumarle la multiplicación de la clase real con el valor del atributo.

Ahora simplemente el entrenamiento consiste en hacer una predicción, si ésta es correcta seguimos con lo siguientes datos, si es incorrecta ajustamos los weights.

````python
def train(self, data, epochs):
        for epoch in range(epochs):
            print('Epoch {}'.format(epoch))
            for datum in data:
                real_class = self.dict_classes[datum[len(datum)-1]]
                prediction_class = self.predict(datum)
                if real_class != prediction_class:
                    self.adjust_weights(real_class,datum)
                    print('Adjusted weights: {}'.format(self.weights))
            print('Final weights from epoch {}: {}'.format(epoch,self.weights))
````

Te darás cuenta de que recibe un atributo epochs, el cual indica cuántas veces iterará la misma información para ajustar cada vez más los weights.

Por último, pero no menos importante, falta el método de verificación, qué es exactamente lo mismo que en el entrenamiento, sólo que aquí ya no se ajustan los weights. El método regresará en porcentaje cuántas predicciones fueron correctas.

````python
def verify(self, data):
        count = 0
        for datum in data:
            real_class = self.dict_classes[datum[len(datum)-1]]
            prediction_class = self.predict(datum)
            if real_class != prediction_class:
                count = count + 1
        return (1-count/len(data))*100
````

Pues ya quedó listo el perceptrón, ahora solo falta utilizarlo, creamos un perceptrón y le daremos como parámetros el número de weights (el cual es 5, porque son 4 atributos + bias), y las clases que hemos cargado, en mi caso yo estoy comparando iris-setosa e iris-versicolor.

````python
perceptron = Perceptron(5, ['virginica', 'versicolor'])
````
