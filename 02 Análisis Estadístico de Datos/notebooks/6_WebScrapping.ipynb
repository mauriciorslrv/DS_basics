## 1. Dos caminos

**Web:** petición → HTML → localizar → extraer → estructurar  
**API:** petición → JSON/XML → validar → transformar

Cuando existe una API oficial adecuada, suele ser más estable que depender de la estructura visual de una página.

## 2. HTML sin depender de internet

Usaremos HTML local para aprender sin que la notebook falle si una web cambia.

```python
from bs4 import BeautifulSoup
import pandas as pd

html = '''
<html><body>
<div class="producto"><h2>Café</h2><span class="precio">120</span></div>
<div class="producto"><h2>Té</h2><span class="precio">85</span></div>
<div class="producto"><h2>Cacao</h2><span class="precio">150</span></div>
</body></html>
'''

soup=BeautifulSoup(html,"html.parser")
filas=[]
for producto in soup.select(".producto"):
    filas.append({
        "producto":producto.select_one("h2").get_text(strip=True),
        "precio":float(producto.select_one(".precio").get_text(strip=True))
    })
df=pd.DataFrame(filas)
df
```

### 💡 IDEA
Scrapear no es "copiar una página": es convertir una estructura semiestructurada en datos validables.
