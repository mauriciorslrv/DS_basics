## 1. De un dato a muchos resultados posibles

Una **variable aleatoria** asigna números a resultados posibles. Una **distribución** describe qué resultados son posibles y qué tan plausibles son.

### 💡 IDEA
La probabilidad no elimina la incertidumbre: la **estructura**.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng=np.random.default_rng(42)
```

## 2. Discreta: Poisson para conteos

```python
k=np.arange(0,13); lam=4
pmf=stats.poisson.pmf(k,mu=lam)
plt.bar(k,pmf); plt.xlabel("Pedidos/min"); plt.ylabel("Probabilidad"); plt.show()
```

## 3. Binomial: éxitos en un número fijo de intentos

```python
n,p=20,.85
x=np.arange(n+1)
plt.bar(x,stats.binom.pmf(x,n=n,p=p))
plt.xlabel("Entregas a tiempo de 20"); plt.ylabel("Probabilidad"); plt.show()
```

## 4. Continua: distribución normal

En variables continuas la probabilidad corresponde a un **intervalo**, no a la altura de un punto.

```python
x=np.linspace(10,65,400); media,sigma=36,7
pdf=stats.norm.pdf(x,loc=media,scale=sigma)
plt.plot(x,pdf); plt.xlabel("Minutos"); plt.ylabel("Densidad"); plt.show()
```

### ⚠️ ERROR ÚTIL
En una distribución continua, `pdf(x)` **no** es la probabilidad de observar exactamente `x`. Para intervalos usamos la CDF.

```python
prob=stats.norm.cdf(45,media,sigma)-stats.norm.cdf(30,media,sigma)
print(f"P(30 <= T <= 45) = {prob:.2%}")
```

## 5. Uniforme y muestreo

```python
m=rng.uniform(0,1,5000)
plt.hist(m,bins=25,density=True); plt.show()
```

## 6. Modelo teórico vs. datos

Que una curva se vea parecida a los datos **no demuestra** que el proceso siga exactamente esa distribución.

```python
muestra=rng.normal(36,7,1200)
plt.hist(muestra,bins=30,density=True,alpha=.6,label="muestra")
plt.plot(x,pdf,label="modelo"); plt.legend(); plt.show()
```
