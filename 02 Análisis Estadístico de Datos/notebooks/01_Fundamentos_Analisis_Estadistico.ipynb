La estadística no empieza con una fórmula. Empieza con una **pregunta**, una población de interés y una decisión sobre qué observar.


## 1. Una pregunta antes del cálculo

Imagina un servicio de entregas que quiere responder:

> *+¿Nuestros tiempos de entrega son consistentes y qué factores parecen relacionarse con retrasos?**

Antes de calcular promedios distinguimos:

- **Población:** todos los casos sobre los que queremos razonar.
- **Muestra:** los casos que realmente observamos.
- **Variable:** una característica medida o registrada.
- **Parámetro:** una cantidad de la población que normalmente desconocemos.
- **Estadístico:** una cantidad calculada a partir de la muestra.

### 💡 IDEA
Una muestra puede informarnos sobre una población, pero la calidad de la conclusión depende también de **cómo fue obtenida**.

## 2. Tipos de variables

| Tipo | Ejemplo | Qué representa |
|---|---|---|
| Nominal | canal = app / web / tienda | categorías sin orden |
| Ordinal | satisfacción = baja / media / alta | categorías con orden |
| Discreta | número de productos | conteos |
| Continua | minutos de entrega | mediciones |

El tipo de variable orienta qué resumen, gráfica o prueba tiene sentido.


### ¿Por qué este ejercicio?
Construiremos un pequeño dataset sintético de entregas porque combina variables **categóricas y numéricas**. La meta no es memorizar `pandas`: es entrenar la decisión de **qué significa cada columna antes de calcular con ella**.


```python
import numpy as np
import pandas as pd

# Usamos una seed fija para que el ejemplo sea reproducible.
rng = np.random.default_rng(42)

entregas = pd.DataFrame({
    "tiempo_min": np.round(rng.normal(36, 7, 40), 1),
    "canal": rng.choice(["app", "web", "tienda"], 40, p=[0.55, 0.30, 0.15]),
    "productos": rng.integers(1, 6, 40),
    "satisfaccion": rng.choice(["baja", "media", "alta"], 40, p=[0.15, 0.45, 0.40])
})

# Esta columna es derivada: el umbral define qué llamamos "retraso".
entregas["retraso"] = entregas["tiempo_min"] > 45
entregas.head()
```

## 3. Describir antes de inferir

Primero resumimos **lo observado**. `describe()` no responde por nosotros; sólo organiza información que después debemos interpretar.


```python
# Resumen para variables numéricas.
entregas[["tiempo_min", "productos"]].describe()
```

```python
# Proporción observada por canal.
entregas["canal"].value_counts(normalize=True).rename("proporcion")
```

### 🔬 PRUEBA
Cambia el umbral de retraso de 45 a 40 minutos.

¿Cambió el dato original o cambió nuestra **definición operacional** de retraso?

Esta distinción aparece constantemente en análisis reales: una métrica depende de cómo definimos lo que medimos.


## 4. Descriptiva vs. inferencial

- **Estadística descriptiva:** resume la muestra observada.
- **Estadística inferencial:** usa la muestra para razonar sobre una población bajo supuestos.

El salto de describir a inferir exige pensar en muestreo, incertidumbre y posibles sesgos.


```python
print(f"Tiempo promedio observado: {entregas['tiempo_min'].mean():.1f} min")
print(f"Proporción observada de retrasos: {entregas['retraso'].mean():.1%}")
```

### ⚠️ ERROR ÚTIL
Encontrar una diferencia entre grupos **no demuestra causalidad**. Puede haber variables no medidas, selección no aleatoria o definiciones que influyan en el resultado.
