## 1. De un dato a muchos resultados posibles

Una **variable aleatoria** asigna números a resultados posibles. Una **distribución** describe qué resultados son posibles y qué tan plausibles son.

### 💡 IDEA
La probabilidad no elimina la incertidumbre: la **estructura**.

En lugar de preguntar sólo "¿cuál será el resultado?", empezamos a preguntar:

> **¿qué resultados pueden ocurrir y con qué frecuencia o probabilidad?**


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

rng=np.random.default_rng(42)
```

## 2. Discreta: Poisson para conteos

### ¿Por qué este ejercicio?
Poisson es una primera forma útil de modelar **conteos de eventos en un intervalo** cuando una tasa promedio tiene sentido.

Ejemplo conceptual: pedidos que llegan por minuto.


```python
k=np.arange(0,13)
lam=4  # tasa media esperada por intervalo

pmf=stats.poisson.pmf(k,mu=lam)

plt.bar(k,pmf)
plt.xlabel("Pedidos por minuto")
plt.ylabel("Probabilidad")
plt.show()
```

## 3. Binomial: éxitos en un número fijo de intentos

### ¿Por qué este ejercicio?
La Binomial sirve cuando tenemos:

1. un número fijo de intentos;
2. dos resultados posibles por intento;
3. una probabilidad de éxito definida bajo el modelo.

Ejemplo: de 20 entregas, ¿cuántas llegan a tiempo?


```python
n,p=20,.85
x=np.arange(n+1)

plt.bar(x,stats.binom.pmf(x,n=n,p=p))
plt.xlabel("Entregas a tiempo de 20")
plt.ylabel("Probabilidad")
plt.show()
```

## 4. Continua: distribución normal

En variables continuas la probabilidad corresponde a un **intervalo**, no a la altura de un único punto.

La normal aparece con frecuencia como modelo aproximado, pero **no debemos asumirla automáticamente**.


```python
x=np.linspace(10,65,400)
media,sigma=36,7

pdf=stats.norm.pdf(x,loc=media,scale=sigma)

plt.plot(x,pdf)
plt.xlabel("Minutos")
plt.ylabel("Densidad")
plt.show()
```

### ⚠️ ERROR ÚTIL
En una distribución continua, `pdf(x)` **no** es la probabilidad de observar exactamente `x`.

Para probabilidades de intervalos usamos la **CDF** (función de distribución acumulada).


```python
prob=stats.norm.cdf(45,media,sigma)-stats.norm.cdf(30,media,sigma)
print(f"P(30 <= T <= 45) = {prob:.2%}")
```

## 5. Uniforme y muestreo

### ¿Por qué este ejercicio?
La uniforme es una buena distribución para desarrollar intuición: dentro del intervalo especificado, todos los valores tienen la misma densidad.

También nos prepara para Monte Carlo, donde generaremos repetidamente valores aleatorios.


```python
m=rng.uniform(0,1,5000)

plt.hist(m,bins=25,density=True)
plt.xlabel("Valor")
plt.ylabel("Densidad")
plt.show()
```

## 6. Modelo teórico vs. datos

Que una curva se vea parecida a los datos **no demuestra** que el proceso siga exactamente esa distribución.

Un modelo es una representación útil bajo supuestos; su calidad depende del propósito y de cuánto se ajuste a la evidencia relevante.


```python
muestra=rng.normal(36,7,1200)

plt.hist(muestra,bins=30,density=True,alpha=.6,label="muestra simulada")
plt.plot(x,pdf,label="modelo normal")
plt.legend()
plt.show()
```
